### Python Program: Contrast Stretching and Bit-Plane Slicing

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("../cat.jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
r_min = np.min(image)
r_max = np.max(image)
```

```python
contrast = (image-r_min)/(r_max-r_min) * 255
```

```python
contrast = contrast.astype(np.uint8)
```

# 2. Bit-Plane Slicing


```python
bit_planes = []
contrast = cv2.cvtColor(contrast, cv2.COLOR_BGR2GRAY)
```

```python
for i in range(8):
    bit_plane = (contrast >> i) & 1
    bit_plane = bit_plane * 255  # Scale to 0-255 for visualization
    bit_plane = bit_plane.astype(np.uint8)
    bit_planes.append(bit_plane)
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Stretched")
plt.axis("off")

plt.tight_layout()
plt.show()


# ==========================================
# Display Bit Planes
# ==========================================

plt.figure(figsize=(12, 8))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_planes[i], cmap="gray")
    plt.title(f"Bit Plane {i}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
