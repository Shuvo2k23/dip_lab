### Python Program: Gray-Level Slicing With and Without Background

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("../cat.jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
r1 = 100
r2 = 200
```

```python
without_bg = np.where((gray >= r1) & (gray <= r2), 255, 0)

with_bg = np.where((gray >= r1) & (gray <= r2), 255, gray)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(without_bg, cmap="gray")
plt.title("Slicing Without Background")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(with_bg, cmap="gray")
plt.title("Slicing With Background")
plt.axis("off")

plt.tight_layout()
plt.show()
```
