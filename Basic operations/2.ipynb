# Python Program: Analyze Effects of Changing Spatial and Intensity Resolution

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("dog.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

# 1. Change Spatial Resolution


```python
sp1 = cv2.resize(img,(512,512))
sp2 = cv2.resize(img,(256,256))
sp3 = cv2.resize(img,(128,128))
sp4 = cv2.resize(img,(64,64))
```

```python
sp2 = cv2.resize(sp2,(512,512),interpolation=cv2.INTER_NEAREST)
sp3 = cv2.resize(sp3,(512,512),interpolation=cv2.INTER_NEAREST)
sp4 = cv2.resize(sp4,(512,512),interpolation=cv2.INTER_NEAREST)
```

# 2. Change Intensity Resolution


```python
def reduce_intensity(image, bits):  
    
    levels = 2 ** bits
    factor = 256 // levels
    reduced_image = (image // factor) * factor
    return reduced_image.astype(np.uint8)
```

```python
in_8bit = reduce_intensity(sp1, 8)
in_6bit = reduce_intensity(sp1, 6)
in_4bit = reduce_intensity(sp1, 4)
in_2bit = reduce_intensity(sp1, 2)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(in_8bit, cmap='gray')
plt.title('8-bit')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(in_6bit, cmap='gray')
plt.title('6-bit')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(in_4bit, cmap='gray')
plt.title('4-bit')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(in_2bit, cmap='gray')
plt.title('2-bit')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(sp1)
plt.title("Spatial Resolution: 512 × 512")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(sp2)
plt.title("Spatial Resolution: 256 × 256")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(sp3)
plt.title("Spatial Resolution: 128 × 128")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(sp4)
plt.title("Spatial Resolution: 64 × 64")
plt.axis("off")

plt.tight_layout()
plt.show()
```
