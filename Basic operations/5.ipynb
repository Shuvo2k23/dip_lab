# Python Program: Color Image Operations

```python
import cv2
import matplotlib.pyplot as plt

# Read color image
image = cv2.imread("dog.jpg")

# Convert BGR to RGB
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


# ------------------------------------------------
# 1. Convert Color Image to Grayscale Image
# ------------------------------------------------

gray = cv2.cvtColor(rgb, cv2.COLOR_RGB2GRAY)


# ------------------------------------------------
# 2. Convert Color Image to Binary Image
# ------------------------------------------------

# First convert to grayscale, then apply threshold
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)


# ------------------------------------------------
# 3. Separate Image into R, G, B Planes
# ------------------------------------------------

red = rgb[:, :, 0]
green = rgb[:, :, 1]
blue = rgb[:, :, 2]


# ------------------------------------------------
# 4. Combine R, G, B Planes to Create Color Image
# ------------------------------------------------

combined = cv2.merge([red, green, blue])


# ------------------------------------------------
# Display Results
# ------------------------------------------------

plt.figure(figsize=(12, 10))

# Original image
plt.subplot(2, 3, 1)
plt.imshow(rgb)
plt.title("Original Color Image")
plt.axis("off")

# Grayscale
plt.subplot(2, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

# Binary
plt.subplot(2, 3, 3)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

# Red plane
plt.subplot(2, 3, 4)
plt.imshow(red, cmap="gray")
plt.title("Red Plane")
plt.axis("off")

# Green plane
plt.subplot(2, 3, 5)
plt.imshow(green, cmap="gray")
plt.title("Green Plane")
plt.axis("off")

# Blue plane
plt.subplot(2, 3, 6)
plt.imshow(blue, cmap="gray")
plt.title("Blue Plane")
plt.axis("off")

plt.tight_layout()
plt.show()


# Display reconstructed image separately
plt.figure(figsize=(5, 5))
plt.imshow(combined)
plt.title("Combined RGB Image")
plt.axis("off")
plt.show()
```
