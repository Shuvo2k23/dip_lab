# Convert Grayscale to Binary

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(rgb_image)
plt.title('RGB Image')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(binary_image,cmap='gray')
plt.title('Binary Image')            
plt.axis('off')
plt.subplot(1,3,3)
plt.imshow(gray_image,cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("dog.jpg")

# Convert BGR to RGB (OpenCV reads images in BGR format)
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert RGB to Grayscale
gray_image = cv2.cvtColor(rgb_image, cv2.COLOR_RGB2GRAY)

# Convert Grayscale to Binary
# Threshold value = 127
binary_image = cv2.threshold(
    gray_image, 127, 255, cv2.THRESH_BINARY
)[1]

# Display images
plt.figure(figsize=(12, 4))

# Original RGB Image
plt.subplot(1, 3, 1)
plt.imshow(rgb_image)
plt.title("RGB Image")
plt.axis("off")

# Grayscale Image
plt.subplot(1, 3, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

# Binary Image
plt.subplot(1, 3, 3)
plt.imshow(binary_image, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
