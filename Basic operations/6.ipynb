# Python Program: Arithmetic and Logical Operations on Images

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img1 = cv2.imread("dog.jpg")
img2 = cv2.imread("cat.jpg")
```

```python
img2 = cv2.resize(img2, (img1.shape[1],img1.shape[0]))
```

```python
img1 = cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
```

```python
add = cv2.add(img1,img2)
```

```python
sub = cv2.subtract(img1,img2)
```

```python
logical_and = cv2.bitwise_and(img1,img2)
logical_or = cv2.bitwise_or(img1,img2)
```

```python
logical_not = cv2.bitwise_not(img1)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(add)
plt.title("Addition")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(sub)
plt.title("Subtraction")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(logical_and)
plt.title("AND")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(logical_or)
plt.title("OR")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(logical_not)
plt.title("NOT")
plt.axis("off")

plt.tight_layout()
plt.show()
```
