# 13. Write a python program to implement non-linear filters (Median, Max, Min filters) for noise reduction.

**Purpose:** Imports the libraries used for image filtering.

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

**Purpose:** Loads the dog image and converts it to grayscale.

```python
img = cv2.imread('../dog.jpg')
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

**Purpose:** Defines a helper that applies median, maximum, or minimum neighborhood filtering.

```python
def apply_filter(img, type):
    h,w = img.shape
    nimg = np.zeros_like(img)
    for i in range(1,h-1):
        for j in range(1,w-1):
            neighbor = img[i-1:i+2,j-1:j+2]
            if type == 'median':
                sum = np.sum(neighbor)
                nimg[i][j] = sum//9
            elif type == 'max':
                mx = np.max(neighbor)
                nimg[i][j] = mx
            else:
                mn = np.min(neighbor)
                nimg[i][j] = mn
    return nimg
```

**Purpose:** Applies the three nonlinear filters to the grayscale image.

```python
mfilter = apply_filter(gray,'median')
maxfilter = apply_filter(gray,'max')
minfilter = apply_filter(gray,'min')
```

**Purpose:** Displays the median-, maximum-, and minimum-filtered images.

```python
plt.figure(figsize=(18,6))
plt.subplot(1,3,1)
plt.imshow(mfilter,cmap='gray')
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(maxfilter,cmap='gray')
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(minfilter,cmap='gray')
plt.axis("off")
```
