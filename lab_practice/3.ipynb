### 3. Write a python program to perform geometric transformations (Resizing/Zooming, Rotation, Translation).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread("../dog.jpg")
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img,cmap='gray')
height,width,dim = img.shape
print(img.shape)
```

**Purpose:** Enlarges the image with linear interpolation and displays it.

```python
zooming = cv2.resize(img,(width*5,height*5),interpolation=cv2.INTER_LINEAR)
plt.imshow(zooming,cmap='gray')
```

**Purpose:** Displays the original image for comparison.

```python
plt.imshow(img,cmap='gray')
```

**Purpose:** Rotates the image by ten degrees and displays it.

```python

center = (height//2,width//2)
rotation_matrix = cv2.getRotationMatrix2D(center=center,angle= 10,scale=1.0)
rotated_img = cv2.warpAffine(img,rotation_matrix,(width,height))
plt.imshow(rotated_img,cmap='gray')
```

**Purpose:** Shifts the image horizontally and vertically and displays it.

```python
tx = 20
ty = 20

import numpy as np

translation_matrix = np.float32([[1,0,tx],
                                [0,1,ty]])
translated = cv2.warpAffine(img,translation_matrix,(width,height))
plt.imshow(translated,cmap='gray')
```
