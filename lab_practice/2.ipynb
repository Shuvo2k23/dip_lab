### Reduce Image size

```python
spatial = image[::4,::4]
spatial.shape
```

```python
sp1 = cv2.resize(image,(128,128),interpolation=cv2.INTER_NEAREST)
plt.imshow(sp1)
```

```python
def reduce_intensity(img, bits):
    lavels = 2**bits
    factor = 255//lavels
    rimg = (image//factor)*factor
    return rimg
```

```python
im_4bit = reduce_intensity(image,5)
plt.imshow(im_4bit, cmap='gray')
```
