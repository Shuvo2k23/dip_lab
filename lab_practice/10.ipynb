**Purpose:** Loads the cat image as grayscale and imports plotting tools.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('../cat.jpg',cv2.IMREAD_GRAYSCALE)
```

**Purpose:** Gets the image height and width.

```python
height , width = img.shape
```

**Purpose:** Counts the frequency of each grayscale intensity.

```python
hist = [0] * 256

for i in range(height):
    for j in range(width):
        pixel = img[i][j]
        hist[pixel] += 1
```

**Purpose:** Builds the cumulative distribution from the histogram.

```python
cdf = [0] * 256
cdf[0] = hist[0]

for i in range(1,256):
    cdf[i] = cdf[i-1] + hist[i]
    
```

**Purpose:** Introduces the histogram equalization section.

## Equalized image

**Purpose:** Maps each pixel through the cumulative distribution to equalize the image.

```python
equalized = np.zeros_like(img)
total_pixel = height* width
for i in range(height):
    for j in range(width):
        pixel = img[i][j]
        equalized[i][j] = int((cdf[pixel]/total_pixel)*255)
```

**Purpose:** Counts the intensity frequencies in the equalized image.

```python
eq_hist = [0] * 256

for i in range(height):
    for j in range(width):
        pixel = equalized[i][j]
        eq_hist[pixel] += 1
```

**Purpose:** Displays the original and equalized images with their histograms.

```python
plt.figure(figsize=(12, 8))
plt .subplot(2, 2, 1)
plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.bar(range(256),hist)
plt.title('Histogram of Original Image')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.subplot(2, 2, 3)
plt.imshow(equalized, cmap='gray')
plt.title('Equalized Image')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.bar(range(256),eq_hist )
plt.title('Histogram of Equalized Image')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
