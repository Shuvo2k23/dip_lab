**Purpose:** Loads the dog image and imports the image-processing libraries.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread("../dog.jpg")
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

plt.imshow(img,cmap='gray')
height,width,dim = img.shape
```

**Purpose:** Separates the red, green, and blue channels and displays one channel.

```python
r_img = img[:,:,0]
g_img = img[:,:,1]
b_img = img[:,:,2]

plt.imshow(r_img,cmap='Blues')
```

**Purpose:** Merges the color channels and displays the reconstructed image.

```python
combined = cv2.merge([r_img,g_img,b_img])
plt.imshow(combined,cmap='gray')
```
