**Purpose:** Imports libraries and loads the dog and cat images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img1 = cv2.imread("../dog.jpg",cv2.COLOR_BGR2RGB)
img2 = cv2.imread("../cat.jpg",cv2.COLOR_BGR2RGB)
```

**Purpose:** Resizes the second image to match the first image dimensions.

```python
img2 = cv2.resize(img2,(img1.shape[1],img1.shape[0]))
```

**Purpose:** This cell is intentionally left blank.

**Purpose:** Adds the two images pixel by pixel.

```python
add = cv2.add(img1,img2)
# print(img2.shape,img1.shape)
```

**Purpose:** Subtracts the first image from the second.

```python
sub = cv2.subtract(img2,img1)
```

**Purpose:** Computes bitwise AND and OR results for the images.

```python
logical_and = cv2.bitwise_and(img1,img2)
logical_or = cv2.bitwise_or(img1,img2)
```

**Purpose:** Inverts the pixel values of the first image.

```python
logical_not = cv2.bitwise_not(img1)
```

**Purpose:** Displays the arithmetic and bitwise image-operation results.

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(add)
plt.title("Addition")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(sub)
plt.title("Subtraction")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(logical_and)
plt.title("AND")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(logical_or)
plt.title("OR")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(logical_not)
plt.title("NOT")
plt.axis("off")

plt.tight_layout()
plt.show()
```
