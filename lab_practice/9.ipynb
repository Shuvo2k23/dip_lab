```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("../cat.jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
r_min = np.min(gray)
r_max = np.max(gray)
```

```python
contrast = (gray-r_min)/(r_max-r_min)*255
```

```python
contrast = contrast.astype(np.uint8)
```

## bit plane slicing

```python
planes = []

for i in range (8):
    bp = (gray>>i) & 1 
    bp = bp* 255
    bp = bp.astype(np.uint8)
    planes.append(bp)
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Stretched")
plt.axis("off")

plt.tight_layout()
plt.show()


# ==========================================
# Display Bit Planes
# ==========================================

plt.figure(figsize=(12, 8))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(planes[i], cmap="gray")
    plt.title(f"Bit Plane {i}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
