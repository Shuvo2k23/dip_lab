## 11. Write a python program to apply log and power-law (gamma) transformations.

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('../cat.jpg',cv2.IMREAD_GRAYSCALE)
```

**Purpose:** Applies a logarithmic intensity transformation.

```python
c = 255 / np.log(1+ np.max(img))

log_img = c * np.log(1 + img)
log_img = log_img.astype(np.uint8)
```

**Purpose:** Introduces the gamma transformation section.

## gamma Transform


**Purpose:** Sets the gamma exponent and scaling constant.

```python
gamma = 2
c = 1
```

**Purpose:** Applies gamma correction to the normalized image.

```python
norm = img/255
gamma_img = c* np.power(norm,gamma)*255
gamma_img = gamma_img.astype(np.uint8)
gamma_img
```

**Purpose:** Displays the original image and the log- and gamma-transformed results.

```python
# =================================
# Display Results
# =================================

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(log_img, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(gamma_img, cmap="gray")
plt.title("Gamma = 0.5")
plt.axis("off")

plt.tight_layout()
plt.show()
```
