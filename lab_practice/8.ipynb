```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

```

```python
image = cv2.imread("../cat.jpg",cv2.IMREAD_GRAYSCALE)

```

```python
r1 = 100
r2 = 200
```

```python
without_bg = np.where((image>=r1) & (image<=r2) , 255, 0)
with_bg = np.where((image>=r1) & (image<=r2 ), 255, image)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(without_bg, cmap="gray")
plt.title("Slicing Without Background")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(with_bg, cmap="gray")
plt.title("Slicing With Background")
plt.axis("off")

plt.tight_layout()
plt.show()
```
