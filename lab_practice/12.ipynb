**Purpose:** Imports the libraries used for mean filtering.

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

**Purpose:** Loads the cat image and converts it to grayscale.

```python
image = cv2.imread('../cat.jpg')

gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

**Purpose:** Gets the grayscale image dimensions.

```python
h,w = gray.shape
```

**Purpose:** Creates a normalized 3-by-3 averaging kernel.

```python
kernel = np.ones((3,3)).astype('float')/9.0
```

**Purpose:** Applies the averaging kernel manually across the image.

```python
filterd = np.zeros_like(gray)

for i in range (1,h-1):
    for j in range(1,w-1):
        neighbour = gray[i-1:i+2,j-1:j+2]
        filterd[i][j] = np.sum(kernel*neighbour)
```

**Purpose:** Displays the original grayscale image.

```python
plt.imshow( gray,cmap='gray')
plt.axis("off")
```

**Purpose:** Displays the mean-filtered image.

```python
plt.imshow(filterd,cmap='gray')
```
